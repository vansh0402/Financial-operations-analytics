```python
print("Accounts:", accounts["account_id"].nunique())
print("Subscriptions:", subscriptions["account_id"].nunique())
print("Churn:", churn["account_id"].nunique())
print("Support:", support["account_id"].nunique())
print("Feature Usage:", feature_usage["subscription_id"].nunique())
```

```python
# ============================================================
# BUSINESS INSIGHTS - KEY KPIs
# ============================================================

# -----------------------------
# 1. Unique Entity Counts
# -----------------------------

total_accounts = accounts["account_id"].nunique()
total_subscribers = subscriptions["account_id"].nunique()
total_churned = churn["account_id"].nunique()
total_support_customers = support["account_id"].nunique()
total_feature_users = feature_usage["subscription_id"].nunique()


# -----------------------------
# 2. Business KPI Calculations
# -----------------------------

# Subscription penetration
subscription_penetration = (
    total_subscribers / total_accounts * 100
    if total_accounts > 0 else 0
)

# Churn rate
churn_rate = (
    total_churned / total_subscribers * 100
    if total_subscribers > 0 else 0
)

# Support contact rate
support_contact_rate = (
    total_support_customers / total_accounts * 100
    if total_accounts > 0 else 0
)

# Feature adoption rate
feature_adoption_rate = (
    total_feature_users / subscriptions["subscription_id"].nunique() * 100
    if subscriptions["subscription_id"].nunique() > 0 else 0
)


# -----------------------------
# 3. Display Results
# -----------------------------

print("=" * 55)
print("           BUSINESS INSIGHTS - KEY KPIs")
print("=" * 55)

print(f"Total Accounts              : {total_accounts:,}")
print(f"Total Subscribers           : {total_subscribers:,}")
print(f"Total Churned Customers     : {total_churned:,}")
print(f"Customers Using Support     : {total_support_customers:,}")
print(f"Subscriptions Using Features: {total_feature_users:,}")

print("-" * 55)

print(f"Subscription Penetration    : {subscription_penetration:.2f}%")
print(f"Churn Rate                  : {churn_rate:.2f}%")
print(f"Support Contact Rate        : {support_contact_rate:.2f}%")
print(f"Feature Adoption Rate       : {feature_adoption_rate:.2f}%")

print("=" * 55)
```

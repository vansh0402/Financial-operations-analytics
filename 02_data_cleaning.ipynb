```python
import pandas as pd
import numpy as np 

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
accounts = pd.read_csv(r'C:\PROJECTS\Financial Operations analytics\Data\Raw\ravenstack_accounts.csv')
subscriptions = pd.read_csv('C:\\PROJECTS\\Financial Operations analytics\\Data\\Raw\\ravenstack_subscriptions.csv')
churn = pd.read_csv('C:\\PROJECTS\\Financial Operations analytics\\Data\\Raw\\ravenstack_churn_events.csv')
feature_usage = pd.read_csv('C:\\PROJECTS\\Financial Operations analytics\\Data\\Raw\\ravenstack_feature_usage.csv')
support = pd.read_csv('C:\\PROJECTS\\Financial Operations analytics\\Data\\Raw\\ravenstack_support_tickets.csv')
```

### Changing Data Types for Time-data 

```python
accounts["signup_date"] = pd.to_datetime(accounts["signup_date"])

subscriptions["start_date"] = pd.to_datetime(subscriptions["start_date"])
subscriptions["end_date"] = pd.to_datetime(subscriptions["end_date"])

feature_usage["usage_date"] = pd.to_datetime(feature_usage["usage_date"])

support["submitted_at"] = pd.to_datetime(support["submitted_at"])
support["closed_at"] = pd.to_datetime(support["closed_at"])

churn["churn_date"] = pd.to_datetime(churn["churn_date"])
```

# Cleaning Summary

## Cleaning Actions Performed
- Converted date columns to datetime format.

### Accounts
- No cleaning was required.
- No missing values or duplicate records were identified.

### Subscription
- Retained missing values in the `end_date` column as they likely represent active subscriptions.
- Created a new `is_active` column based on the availability of `end_date`.

### Feature Usage
- No cleaning was required.
- The dataset contained no missing values or duplicate records.

### Support Tickets

- Retained missing values in `satisfaction_score` since customer feedback is optional.

### Churn Events
- The `reason_code` value "unknown" was retained because it represents a valid business category rather than missing data.
- Missing values in `feedback_text` were retained since customer feedback is optional and the absence of feedback is meaningful.
